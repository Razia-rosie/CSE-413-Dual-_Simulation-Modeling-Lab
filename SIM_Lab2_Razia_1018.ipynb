

## Task 1 & 2: Create Personal Matrix A and Analyze Matrix A

```python
import numpy as np

# Student ID: 18
# Last digit = 8, second-last digit = 1
d1 = 8
d2 = 1

# Personal Matrix A = [[d1+2, d2+1], [2*d1, d2+2]]
A = np.array([[d1 + 2, d2 + 1],
              [2 * d1, d2 + 2]])

print("Matrix A:\n", A)

# Shape
print("\nShape of A:", A.shape)

# Determinant
det_A = np.linalg.det(A)
print("Determinant of A:", det_A)

# Rank
rank_A = np.linalg.matrix_rank(A)
print("Rank of A:", rank_A)

# Eigenvalues
eigen_A = np.linalg.eigvals(A)
print("Eigenvalues of A:", eigen_A)

# Inverse
if not np.isclose(det_A, 0):
    inverse_A = np.linalg.inv(A)
    print("Inverse of A:\n", inverse_A)
else:
    print("A has no inverse")
```

## Task 3 & 4: Create New Matrix B and Re-analyze Matrix B

```python
# Change exactly one value by +1
# A[0,0] changes from 10 to 11
B = A.copy()
B[0, 0] = B[0, 0] + 1

print("Changed element: A[0,0] = 10 to B[0,0] = 11")
print("\nMatrix B:\n", B)

# Shape
print("\nShape of B:", B.shape)

# Determinant
det_B = np.linalg.det(B)
print("Determinant of B:", det_B)

# Rank
rank_B = np.linalg.matrix_rank(B)
print("Rank of B:", rank_B)

# Eigenvalues
eigen_B = np.linalg.eigvals(B)
print("Eigenvalues of B:", eigen_B)

# Inverse
if not np.isclose(det_B, 0):
    inverse_B = np.linalg.inv(B)
    print("Inverse of B:\n", inverse_B)
else:
    print("B has no inverse")
```

## Task 5: Comparison & Explanation

### Matrix A
- Determinant: **-2.0**
- Rank: **2**
- Eigenvalues: approximately **[13.15206735, -0.15206735]**
- Inverse: **[[-1.5, 1.0], [8.0, -5.0]]**

### Matrix B
- Determinant: **1.0**
- Rank: **2**
- Eigenvalues: approximately **[13.92820323, 0.07179677]**
- Inverse: **[[3.0, -2.0], [-16.0, 11.0]]**

### 1. How did the determinant change and why?

The determinant underwent a noticeable shift, moving from **-2.0** for Matrix A to **1.0** for Matrix B – a clear increase of **3.0**. This change makes perfect sense when we look at how Matrix B was created: we incremented the `A[0,0]` element by 1. In a 2x2 matrix, the determinant is calculated as `ad - bc`. By increasing 'a' (the `A[0,0]` element) by 1, the `ad` part of the formula effectively increased by `1 * d` (which is `1 * A[1,1]`, or `1 * 3 = 3`). This direct impact on the 'a' term directly led to the determinant increasing by 3.

### 2. Did the rank change? Explain.

Interestingly, the rank remained consistent, staying at **2** for both matrices. This is because for a 2x2 matrix, having a non-zero determinant is the key indicator of full rank. Since both Matrix A's determinant ( -2.0 ) and Matrix B's determinant ( 1.0 ) are distinctly not zero, both matrices retain their full rank of 2. They are both full rank and thus linearly independent.

### 3. How did the eigenvalues respond to the value change?

The eigenvalues responded quite dynamically to the small change in the `A[0,0]` element. For Matrix A, they were roughly **13.1521** and **-0.1521**. After the change, for Matrix B, they became approximately **13.9282** and **0.0718**. We can observe a couple of things: the larger eigenvalue increased, and the smaller eigenvalue not only increased but also crossed over from negative to a small positive value. This shows how sensitive eigenvalues can be to even minor adjustments within a matrix.

### 4. Is B easier or harder to invert than A? Why?

Both matrices A and B are perfectly invertible because, as we've established, neither has a determinant of zero. However, when we consider numerical stability and ease of computation, Matrix B might be considered slightly 'easier' to invert. Its determinant is **1**, which is often seen as ideal for inversion as it avoids scaling issues with very large or very small determinants. Matrix A's determinant is **-2**, which is also fine but can sometimes lead to slightly larger numbers in the inverse calculation. That said, for simple 2x2 matrices like these, the actual computational complexity of finding the inverse is practically the same for both.
