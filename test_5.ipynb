```python
import numpy as np
import scipy.stats as stats

# 1. Your Student ID's last two digits
d = 18
print(f"1. Student ID's last two digits (d): {d}\n")

# 2. Calculated Observed Frequencies (O)
# Exact formulas from user specification:
# C: 20 + (d % 20) + d
# C++: 25 + (d % 10)
# Java: 30 + (d % 7)
# Python: 35 + (d % 5)

obs_C = 20 + (d % 20) + d
obs_Cpp = 25 + (d % 10)
obs_Java = 30 + (d % 7)
obs_Python = 35 + (d % 5)

observed = np.array([obs_C, obs_Cpp, obs_Java, obs_Python])
categories = ['C', 'C++', 'Java', 'Python']

print("2. Calculated Observed Frequencies (O):")
for cat, obs in zip(categories, observed):
    print(f"   - {cat}: {obs}")
print()

# 3 & 4. Hypotheses
print("3. Null Hypothesis (H0):")
print("   H0: Students have equal preferences for the four programming languages.")
print("4. Alternative Hypothesis (H1):")
print("   H1: Students do not have equal preferences for the four programming languages.\n")

# 5. Expected Frequency (E) for each category
total_N = np.sum(observed)
expected_val = total_N / 4
expected = np.array([expected_val] * 4)
print(f"5. Expected Frequency (E) for each category (Total N = {total_N}):")
for cat in categories:
    print(f"   - {cat}: {expected_val:.2f}")
print()

# 6. Calculated Chi-Square statistic
chi2_stat = np.sum((observed - expected) ** 2 / expected)
print(f"6. Calculated Chi-Square statistic: {chi2_stat:.4f}\n")

# 7. Degrees of Freedom
dof = len(observed) - 1
print(f"7. Degrees of Freedom (df): {dof}\n")

# 8. Critical Value or p-value (at 5% significance level, alpha = 0.05)
alpha = 0.05
critical_value = stats.chi2.ppf(1 - alpha, dof)
p_value = 1 - stats.chi2.cdf(chi2_stat, dof)
print(f"8. Critical Value (at alpha = {alpha}): {critical_value:.4f}")
print(f"   p-value: {p_value:.6f}\n")

# 9. Your final decision
decision = "Reject H0" if chi2_stat > critical_value else "Fail to reject H0"
print(f"9. Your final decision: {decision}\n")

# 10. Brief conclusion in the context of the scenario
if decision == "Reject H0":
    conclusion = "There is sufficient evidence at the 5% significance level to conclude that students do not have equal preferences for the four programming languages."
else:
    conclusion = "There is insufficient evidence at the 5% significance level to suggest that students' preferences for the four programming languages are different."
print(f"10. Brief conclusion:\n    {conclusion}")
```

At the 5% significance level, we reject the null hypothesis because the calculated Chi-Square statistic (8.5652) is greater than the critical value (7.8147), and the p-value (0.035667) is less than 0.05. Therefore, there is sufficient statistical evidence to conclude that students do not have equal preferences for the four programming languages: C, C++, Java, and Python.
