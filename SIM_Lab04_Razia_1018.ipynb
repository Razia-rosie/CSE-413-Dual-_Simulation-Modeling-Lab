# Monte Carlo Simulation for Pi Estimation

```python
import random

def estimate_pi(num_points):
    inside = 0

    for i in range(num_points):
        x = random.random()
        y = random.random()
        if x*x + y*y <= 1:
            inside += 1

    pi = 4 * inside / num_points
    print("Estimated pi for", num_points, "points is:", pi)

estimate_pi(1000)
estimate_pi(5000)
estimate_pi(10000)
estimate_pi(50000)
```

## Part 1: Setup & Personalization Parameters


```python
import random
import numpy as np
import math
import pandas as pd

student_id = "1018"
full_name = "Razia Khanom"

R = int(student_id[-2:])

range_limit = R / 10.0
circle_radius = R / 18.0

num_letters_in_name = sum(c.isalpha() for c in full_name)
min_points_base = 200 * num_letters_in_name

print(f"Student ID Last Two Digits (R): {R}")
print(f"Random Number Range: [0, {range_limit}]")
print(f"Circle Radius: {circle_radius}")
print(f"Number of Letters in Full Name: {num_letters_in_name}")
print(f"Minimum Base Points to Generate: {min_points_base}")
```

## Part 2: Simulation Logic


```python
def is_inside_circle(x, y, radius):
    """Checks if a point (x, y) is inside a circle centered at the origin."""
    return (x**2 + y**2) <= (radius**2)

def run_monte_carlo_pi(total_points, range_limit, radius):
    """Runs a Monte Carlo simulation to estimate Pi."""
    points_inside_circle = 0

    for _ in range(total_points // 2):
        x = random.uniform(0, range_limit)
        y = random.uniform(0, range_limit)
        if is_inside_circle(x, y, radius):
            points_inside_circle += 1

    x_np = np.random.rand(total_points - (total_points // 2)) * range_limit
    y_np = np.random.rand(total_points - (total_points // 2)) * range_limit

    for i in range(len(x_np)):
        if is_inside_circle(x_np[i], y_np[i], radius):
            points_inside_circle += 1

    points_outside_circle = total_points - points_inside_circle

    estimated_pi = 4 * (points_inside_circle / total_points) * ((range_limit**2) / (radius**2))

    error = abs(estimated_pi - math.pi)

    return {
        'Range Used': f"[0, {range_limit}]",
        'Total Points': total_points,
        'Points Inside': points_inside_circle,
        'Estimated π': estimated_pi,
        'Error from math.pi': error
    }
```

## Part 3: Output & Analysis


```python
results = []

for i in range(1, 6):
    current_points = min_points_base * i
    result = run_monte_carlo_pi(current_points, range_limit, circle_radius)
    results.append(result)

df_results = pd.DataFrame(results)
print("Monte Carlo Pi Estimation Results")
display(df_results)
```
