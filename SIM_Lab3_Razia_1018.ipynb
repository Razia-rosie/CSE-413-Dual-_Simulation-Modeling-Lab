## Customer Service Wait Time Simulation

### Part 1 — Data Generation

```python
import numpy as np

# 1. Generate 3,000 wait times using a uniform distribution between 2 and 12 minutes.
regular_day_wait_times = np.random.uniform(low=2, high=12, size=3000)

# 2. Generate 3,000 wait times using a normal distribution with mean=8 and std=2.
busy_day_wait_times = np.random.normal(loc=8, scale=2, size=3000)

# 3. Ensure no negative wait times in the normal distribution (replace negatives with 0).
busy_day_wait_times[busy_day_wait_times < 0] = 0

print("Regular Day Wait Times (first 5):", regular_day_wait_times[:5])
print("Busy Day Wait Times (first 5):", busy_day_wait_times[:5])
```

### Part 2 — Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

# Histogram for Regular Day Wait Times (Uniform Distribution)
sns.histplot(regular_day_wait_times, bins=30, kde=True, color='skyblue', label='Regular Day (Uniform)', alpha=0.7)

# Histogram for Busy Day Wait Times (Normal Distribution)
sns.histplot(busy_day_wait_times, bins=30, kde=True, color='salmon', label='Busy Day (Normal)', alpha=0.7)

plt.title('Distribution of Customer Wait Times')
plt.xlabel('Wait Time (minutes)')
plt.ylabel('Frequency')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

### Part 3 — Statistical Analysis

```python
import numpy as np
import pandas as pd

def analyze_wait_times(data, description):
    mean_val = np.mean(data)
    median_val = np.median(data)
    std_dev_val = np.std(data)
    max_val = np.max(data)

    print(f"--- {description} ---")
    print(f"Mean: {mean_val:.2f} minutes")
    print(f"Median: {median_val:.2f} minutes")
    print(f"Standard Deviation: {std_dev_val:.2f} minutes")
    print(f"Maximum Wait Time: {max_val:.2f} minutes\n")

analyze_wait_times(regular_day_wait_times, "Regular Day Wait Times")
analyze_wait_times(busy_day_wait_times, "Busy Day Wait Times")

# For a more structured output, using pandas DataFrames:
stats_data = {
    'Metric': ['Mean', 'Median', 'Standard Deviation', 'Maximum Wait Time'],
    'Regular Day': [
        np.mean(regular_day_wait_times),
        np.median(regular_day_wait_times),
        np.std(regular_day_wait_times),
        np.max(regular_day_wait_times)
    ],
    'Busy Day': [
        np.mean(busy_day_wait_times),
        np.median(busy_day_wait_times),
        np.std(busy_day_wait_times),
        np.max(busy_day_wait_times)
    ]
}

stats_df = pd.DataFrame(stats_data)
stats_df = stats_df.set_index('Metric')
stats_df = stats_df.round(2) # Round to 2 decimal places for better readability
display(stats_df)
```
